```python
import pandas as pd
```

```python
path = "../data/maestro-v3.0.0-midi/maestro-v3.0.0/maestro-v3.0.0.csv"
df = pd.read_csv(path)
```

```python
composer = df['canonical_composer'].unique()
print(composer)
```

```python
df_chopin = df[df['canonical_composer'] =='Frédéric Chopin']
```

```python
df_chopin.to_csv('../data/processed/chopin.csv', index=False)
```

```python
df_not_chopin = df[df['canonical_composer'] != 'Frédéric Chopin']
```

```python
df_not_chopin.to_csv('../data/processed/not_chopin.csv', index=False)
```

```python
import shutil
import os

destination_directory = "../data/processed/chopin_midi"
os.makedirs(destination_directory,exist_ok=True)
source_directory = "../data/maestro-v3.0.0-midi/maestro-v3.0.0"
count = 0
for index, row in df_chopin.iterrows():
    original_path = os.path.join(source_directory,row['midi_filename'])
    shutil.copy(original_path, destination_directory)
    count = count + 1
print(count)
 
```

```python
import shutil
import os

destination_directory = "../data/processed/other_midi"
os.makedirs(destination_directory,exist_ok=True)
source_directory = "../data/maestro-v3.0.0-midi/maestro-v3.0.0"
count = 0
for index, row in df_not_chopin.iterrows():
    original_path = os.path.join(source_directory,row['midi_filename'])
    shutil.copy(original_path, destination_directory)
    count = count + 1
print(count)
```
