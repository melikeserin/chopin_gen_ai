```python
from miditok import REMI, TokenizerConfig
from pathlib import Path

config = TokenizerConfig(
    num_velocities=32,
    use_chords=True,
    use_tempos=True,
    num_tempos=32,
    use_rests=True,
    use_time_signatures=True,
    additional_tokens={"Chord": True, "Rest": True, "Tempo": True}    
)
tokenizer = REMI(config)
midi_paths =list(Path("../data/processed/chopin_midi").glob("*midi"))
tokenizer.learn_bpe(
    vocab_size = 5000,
    files_paths = midi_paths
)
tokenizer.save_params(Path("../data/processed/chopin_tokenizer.json"))

output_path = "../data/processed/chopin_tokens"
tokenizer.tokenize_midi_dataset( midi_paths,output_path
)
```
